```python
font='DoHyeon_Regular.ttf'
```

```python
moon=open('/content/drive/MyDrive/Colab Notebooks/speech_moon.txt',encoding='UTF-8').read()
moon
```

```python
import re                                 #불필요한 문자 제거, [^가-힣]은 한글이 아닌 모든 문자를 의미하는 정규 표현식임
moon=re.sub('[^가-힣]',' ',moon)
moon
```

```python
import konlpy                                                         #hannanum 만듦
hannanum = konlpy.tag.Hannanum()
hannanum.nouns("대한민국의 영토는 한반도와 그 부속도서로 한다")       #명사 추출
```

```python
nouns=hannanum.nouns(moon)                         #연설문에서 명사 추출
nouns
```

```python
df_word=pd.DataFrame({'word':nouns})
df_word
```

```python
df_word=df_word.groupby('word',as_index=False).agg(n=('word','count')).sort_values('n',ascending=False)
df_word
```

```python
dic_word=df_word.set_index('word').to_dict()['n']        #데이터 프레임을 딕셔너리로
dic_word
```

```python
from wordcloud import WordCloud
wc=WordCloud(random_state=1234,             #난수고정
             font_path=font,                #폰트설정
             width=400,                     #가로크기
             height=400,                    #세로크기
             background_color='white')      #배경색
```

```python
df=pd.read_csv('/content/drive/MyDrive/Colab Notebooks/news_comment_BTS.csv',encoding='UTF-8')
df.info()
```

```python
df['reply']=df['reply'].str.replace('[^가-힣]',' ',regex=True)
df['reply'].head()
```

```python
import konlpy    #kkma만들기
kkma=konlpy.tag.Kkma()
```

```python
nouns=df['reply'].apply(kkma.nouns)       #apply를 활용해 명사 추출
nouns
```

```python
nouns=nouns.explode()
nouns
```

```python
df_word=pd.DataFrame({'word':nouns})                    #데이터 프레임 만들기
df_word['count']=df_word['word'].str.len()              #글자 수 추가
df_word=df_word.query('count >=2')                      #두 글자 이상 단어만 남김
df_word
```

```python
df_word=df_word.groupby('word',as_index=False).agg(n=('word','count')).sort_values('n',ascending=False)
df_word        #빈도표 만들기
```

```python
top20=df_word.head(20)        #단어 빈도 상위 20개 추출
top20
```

```python
plt.rcParams.update({'figure.figsize':[6.5,6]})
sns.barplot(data=top20,y='word',x='n',hue='word')
```

```python
import  folium
folium.Map(location=[35.95, 127.7],                         #지도 중심좌표
           zoom_start=8)                                                #확대 단계
```

```python
map_sig=folium.Map(location=[35.95, 127.7],        #지도 중심좌표
                   zoom_start=8,                   #확대 단계
                   tiles='cartodbpositron')        #지도 종류
map_sig
```

```python
folium.Choropleth(
    geo_data=geo,                                                     #지도 데이터
    data=df_pop,                                                      #통계 데이터
    columns=('code','pop'),                                           #df_pop 행정구역코드, 인구
    key_on='feature.properties.SIG_CD').add_to(map_sig)               #geo 행정 구역 코드
map_sig
```

```python
bins=list(df_pop['pop'].quantile([0,0.2,0.4,0.6,0.8,1]))
bins
```

```python
map_sig=folium.Map(location=[35.95, 127.7],       #지도 중심 좌표
                   zoom_start=8,                  #확대 단계
                   tiles='cartodbpositron')       #지도 종류
```

```python
folium.Choropleth(
    geo_data=geo,                         #지도데이터
    data=df_pop,                          #통계데이터
    columns=('code','pop'),               #df_pop 행정구역코드, 인구
    key_on='feature.properties.SIG_CD',   #geo 행정구역코드
    fill_color='YlGnBu',                  #컬러맵, 소문자 l임()
    fill_opacity=0.7,                     #투명도
    line_opacity=0.5,                     #경계선 투명도
    bins=bins).add_to(map_sig)            #계급 구간기준값, #배경 지도에 추가
map_sig
```

```python
foreigner=pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Foreigner_EMD_Seoul.csv')
foreigner.head()
```

```python
bins=list(foreigner['pop'].quantile([0,0.2,0.4,0.5,0.6,0.7,0.8,0.9,1]))
bins
```

```python
#배경 지도 만들기
map_seoul=folium.Map(location=[37.56,127],            #서울 좌표
                     zoom_start=12,                   #확대 단계
                     tiles='cartodbpositron')         #지도 종류
```

```python
folium.Choropleth(
    geo_data=geo_seoul,                     #지도 데이터
    data=foreigner,                         #통계 데이터
    columns=('code','pop'),                 #foreigner 행정구역 코드, 인구
    key_on='feature.properties.ADM_DR_CD',  #geo_seoul 행정구역 코드
    fill_color='Blues',                     #컬러맵
    nan_fill_color='White',                 #결측치 색깔
    fill_opacity=1,                         #투명도
    line_opacity=0.5,                       #경계선 투명도
    bins=bins).add_to(map_seoul)            #계급 구간 기준값, #배경 지도에 추가
map_seoul
```
